# **Bubble Sort**

Algoritma pengurutan yang pertama adalah *Bubble Sort*. Algoritma ini dikenal sebagai algoritma
pengurutan yang membutuhkan waktu paling lama, hanya saja algoritma ini adalah algoritma
pengurutan yang paling sederhana.

*Bubble sort* mengurutkan data dengan cara sebagai berikut :
1. data dibaca mulai dari paling kiri atau mulai dari indeks ke 0 (phyton) dari suatu list data
2. Bandingkan antara dua buah data yang saling berdekatan (antara data ke i dan i+1), tukar
posisi jika data ke-i lebih besar dibandingkan data ke-i+1 (pengurutan secara ascending)
3. pindah satu posisi
4. lakukan perbandingan seperti langkah ke-2 dan pindah satu posisi kembali seperti langkah
ke-3. Lakukan terus menerus sampai indeks data terakhir.
5. Setelah langkah ke-4 dilakukan, maka data yang berada di posisi paling kanan (indeks ke-N)
adalah data terbesar .
6. Ulangi lagi langkah 1-4 dimulai dari indeks data ke 0 sampai dengan indeks ke N-1, N-2, ..., 0

Setiap iterasi berakhir, data yang terletak di posisi paling kanan adalah data yang paling besar,
oleh karena itu algoritma ini disebut dengan 'bubble' karena data terbesar disetiap iterasi berakhir
bergerak keatas seperti halnya gelembung. Berikut ilustrasi algoritma bubble sort pada tiap iterasi.
Misalkan terdapat data [7, 2.5, 5. 10, 1].

**Iterasi Pertama**, bertujuan untuk memindahkan data terbesar, yaitu 10, ke posisi terakhir atau
paling kanan dari list (karena proses pengurutan secara ascending (dari kecil-ke besar). Proses
iterasi pertama ini dapat dilihat pada Gambar 1.

**Gambar 1. Iterasi Pertama Bubble Sort**

**Iterasi Kedua**, bertujuan untuk memindahkan data terbesar kedua, yaitu 7, agar menempati posisi
tepat sebelum data 10, seperti yang terlihat pada Gambar 2.

**Gambar 2. Iterasi Kedua Bubble Sort**

**Iterasi Ketiga**, bertujuan untuk memindahkan data terbesar ketiga, yaitu 5, untuk menempati
posisi tepat sebelum data 7, seperti yang ditunjukkan pada Gambar 3.

**Gambar 3. Iterasi Ketiga Bubble Sort**

Proses ini dilakukan terus menerus sampai tidak ada data lagi yang akan dibandingkan. Data
dalam keadaan terurut, dapat dilihat pada Gambar 4.

**Gambar 4. Iterasi Terakhir Bubble Sort**

## **Code**

Berikut implementsi lgoritma *bubble sort*.

```python
def bubbleSort(listData):
    print('Data yang akan diurutkan : ', listData)
    count=0
    for outIter in range(len(listData)-1,-1,-1):
        print('outIter=',outIter)
        # count=count+1
        print ('Iterasi ke-', count,':')
        for i in range(outIter):
            if listData[i]>listData[i+1]:
                # temp=listData[i]
                # listData[i]=listData[i+1]
                # listData[i+1]=temp
                listData[i],listData[i+1]=listData[i+1],listData[i]
                count+=1
            print(outIter,'=',listData)
        # print("Jumlah Swap =",count)
    print('Data urut-',listData)
    # print("Total Swap =",count)
```

```python
a=[12,0,5,1,11,20,4,2]
bubbleSort(a)
# b=[12,11,5,1,0]
# print('--')
# bubbleSort(b)
```

```python
b=[11,12,1,5,0]
bubbleSort(b)
```

```python
b=[10,2,5,8,1,20,2,2,4]
bubbleSort(b)
```

1. Pada contoh pengurutan diatas (data b), urutan data hasil dari empat iterasi tidak mengalami perubahan apapun (karena data sudah dalam keaadaan terurut pada setelah iterasi ke-4. Modifikasi algoritma bubble sort tersebut agar iterasi terhenti ketika tidak ada lagi data yang akan ditukar posisinya (semua data sudah berada di tempat yang tepat).

```python
def optimizedBubbleSort(listData):
    print('Data yang akan diurutkan :', listData)
    count = 0
    
    # Perulangan mundur dari index terakhir hingga 0
    for outIter in range(len(listData)-1, -1, -1):
        count += 1
        print('Iterasi ke-', count, ':')
        
        tukar = False  # Flag penanda pertukaran di set False pada awal tiap iterasi
        
        # Perulangan untuk membandingkan data yang berdekatan
        for i in range(outIter):
            if listData[i] > listData[i+1]:
                # Tukar posisi jika data sebelah kiri lebih besar
                listData[i], listData[i+1] = listData[i+1], listData[i]
                tukar = True  # Ubah flag menjadi True karena ada pertukaran
                
        print(f"outIter {outIter} = {listData}")
        
        # Pengecekan Optimasi: Jika tidak ada pertukaran sama sekali, hentikan iterasi
        if not tukar:
            print("--> Tidak ada pertukaran lagi, iterasi dihentikan.")
            break
            
    print('Data urut -', listData)

# --- Contoh Penggunaan ---
# Menggunakan data dari contoh dokumen yang sudah hampir terurut
b = [11, 12, 1, 5, 0] 
optimizedBubbleSort(b)
```

2. Buat modifikasi algoritma bubble sort. Jika pada algoritma bubble sort, pointer bergerak dari kiri ke kanan (menuju index terakhir), sedemikian hingga data terbesar berada pada index terakhir. Modifikasi yang dilakukan adalah, setelah data terbesar terletak pada index terakhir. Pointer bergerak lagi dengan arah sebaliknya, yaitu dari kanan ke kiri untuk membawa data terkecil ke index paling awal (bidirectional bubble sort)

```python
def bidirectionalBubbleSort(listData):
    print('Data yang akan diurutkan :', listData)
    
    kiri = 0
    kanan = len(listData) - 1
    count = 0
    
    while kiri < kanan:
        count += 1
        print(f'\nIterasi Bolak-Balik ke-{count}:')
        tukar = False
        
        # 1. Bergerak dari KIRI ke KANAN (Membawa nilai TERBESAR ke kanan)
        for i in range(kiri, kanan):
            if listData[i] > listData[i+1]:
                # Tukar posisi
                listData[i], listData[i+1] = listData[i+1], listData[i]
                tukar = True
        
        kanan -= 1  # Kurangi batas kanan karena elemen paling kanan sudah pasti yang terbesar
        print(f'  Arah Kiri->Kanan: {listData}')
        
        # Jika tidak ada pertukaran saat bergerak ke kanan, data sudah urut
        if not tukar:
            break
            
        tukar = False # Reset flag untuk pergerakan arah sebaliknya
        
        # 2. Bergerak dari KANAN ke KIRI (Membawa nilai TERKECIL ke kiri)
        # Menggunakan langkah -1 untuk perulangan mundur
        for i in range(kanan, kiri, -1):
            # Bandingkan dengan elemen sebelumnya (i-1)
            if listData[i] < listData[i-1]:
                # Tukar posisi
                listData[i], listData[i-1] = listData[i-1], listData[i]
                tukar = True
                
        kiri += 1  # Tambahkan batas kiri karena elemen paling kiri sudah pasti yang terkecil
        print(f'  Arah Kanan->Kiri: {listData}')
        
        # Jika tidak ada pertukaran saat bergerak ke kiri, data sudah urut
        if not tukar:
            break
            
    print('\nData urut -', listData)

# --- Contoh Penggunaan ---
# Menggunakan contoh array acak
c = [10, 2, 5, 8, 1, 20, 2, 2, 4]
bidirectionalBubbleSort(c)
```
